This notebook demonstrates BaGGLS usage on simulated data with known ground truth effects.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import gamma, bernoulli, norm
import sys
from pathlib import Path

project_root = Path('.').resolve().parent.parent
sys.path.insert(0, str(project_root))

from src.baggls import BaGGLS

np.random.seed(42)

from src.simulation.simulation import *
```

## Generate Simulation Data

```python
# Generate data
n, d = 500, 10
motifs, y, beta_true, design_matrix = generate_simulation_data(n, d, seed=42)

print(f"Generated data: n={n}, d={d}")
print(f"Positive samples: {y.sum()} out of {n}")
print(f"Design matrix shape: {design_matrix.shape}") 
print(f"True non-zero effects: {np.sum(beta_true != 0)}")
```

## Fit BaGGLS Model

```python
# Fit BaGGLS
baggls = fit_baggls_model(y, motifs, verbose=True)
```

## Compute Posterior Estimates and Evaluate Performance

```python
# Compute posterior mean of beta
beta_hat = compute_posterior_mean(baggls)

# Calculate RMSE
rmse = evaluate_rmse(beta_hat, beta_true)

print(f"\nResults:")
print(f"  RMSE: {rmse:.4f}")
print(f"  True non-zero effects: {np.sum(beta_true != 0)}")
print(f"  Estimated effects (|beta| > 0.1): {np.sum(np.abs(beta_hat) > 0.1)}")
```
